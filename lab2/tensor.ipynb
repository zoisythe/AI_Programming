```python
import torch
# 创建一个需要计算梯度的张量
x = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
c = torch.tensor([1.0, 1.0, 1.0], requires_grad=True)
# 定义一个简单的计算图
y = (x ** 2).mean()
#y = (torch.sigmoid(x) ** 2+ (x-c) **2).mean()
# 计算梯度
y.backward()
# 获取梯度
print("张量 x 的梯度:", x.grad)
print("张量 c 的梯度:", c.grad)
```

```python
import torch

# 检查是否有可用的 CUDA 设备
if torch.cuda.is_available():
    device = torch.device("cuda")
    print("CUDA 设备可用，使用 GPU 进行计算。")
else:
    device = torch.device("cpu")
    print("CUDA 设备不可用，使用 CPU 进行计算。")

# 创建一个张量，默认在 CPU 上
x_cpu = torch.tensor([1.0, 2.0, 3.0], requires_grad=True)
print(f"张量 x_cpu 所在设备: {x_cpu.device}")

# 将张量从 CPU 移动到 GPU
x_gpu = x_cpu.to(device)
print(f"张量 x_gpu 所在设备: {x_gpu.device}")

# 或者使用 .cuda() 方法（仅当设备是 GPU 时有效）
if device.type == 'cuda':
    x_gpu_alt = x_cpu.cuda()
    print(f"张量 x_gpu_alt 所在设备: {x_gpu_alt.device}")

# 将张量从 GPU 移动到 CPU
x_cpu_back = x_gpu.cpu()
print(f"张量 x_cpu_back 所在设备: {x_cpu_back.device}")

# 创建一个张量并直接放在 GPU 上
y_gpu = torch.tensor([4.0, 5.0, 6.0], requires_grad=True, device=device)
print(f"张量 y_gpu 所在设备: {y_gpu.device}")

```
